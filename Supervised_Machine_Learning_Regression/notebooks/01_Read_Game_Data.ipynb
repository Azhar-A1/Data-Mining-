```python
from pathlib import Path
import sqlite3
import pandas as pd
PROJECT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA_DIR = PROJECT_DIR / 'data'
games = pd.read_csv(DATA_DIR / 'game_sales_regression.csv')
feedback = pd.read_csv(DATA_DIR / 'player_feedback.csv')
print('Ukuran game:', games.shape, '| feedback:', feedback.shape)
display(games.head())
```

## Baca SQLite

```python
with sqlite3.connect(DATA_DIR / 'game_sales.db') as con:
    display(pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name", con))
    display(pd.read_sql_query('SELECT * FROM games LIMIT 5', con))
```
