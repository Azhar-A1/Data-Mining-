# 03 — Train/Test Split & Transformasi Target

```python
from pathlib import Path
import pandas as pd
PROJECT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
games = pd.read_csv(PROJECT_DIR / 'data' / 'game_sales_regression.csv')
features = ['rating','price_usd','review_count','marketing_budget_musd','release_year']
X = games[features]
y = games['sales_millions']
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
raw=LinearRegression().fit(X_train,y_train).predict(X_test)
log_pred=np.expm1(LinearRegression().fit(X_train,np.log1p(y_train)).predict(X_test))
for name,p in [('Target asli',raw),('Target log1p',log_pred)]:
 print(name,'| MAE:',round(mean_absolute_error(y_test,p),3),'| RMSE:',round(mean_squared_error(y_test,p)**.5,3),'| R2:',round(r2_score(y_test,p),3))
```
