# 03 — Feature Engineering untuk Data Game

Latihan ini membuat fitur sederhana dari data katalog game. Kita belum membangun model machine learning; fokusnya menyiapkan kolom agar lebih mudah dianalisis.

```python
from pathlib import Path
import pandas as pd

PROJECT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
games = pd.read_csv(PROJECT_DIR / 'data' / 'games_eda.csv')
games.head()
```

## 1. Membuat fitur umur game

```python
current_year = 2026
games['game_age'] = current_year - games['release_year']
games[['title','release_year','game_age']].head(10)
```

## 2. Mengelompokkan rating

```python
def rating_category(rating):
    if rating >= 8:
        return 'High'
    elif rating >= 6:
        return 'Medium'
    return 'Low'

games['rating_category'] = games['rating'].apply(rating_category)
display(games[['title','rating','rating_category']].head(10))
print(games['rating_category'].value_counts())
```

## 3. Encoding kategori

```python
games_encoded = pd.get_dummies(games, columns=['genre','platform'], dtype=int)
display(games_encoded.head())
print('Kolom awal:', len(games.columns))
print('Kolom setelah one-hot encoding:', len(games_encoded.columns))
```

## 4. Menyimpan data hasil olahan

```python
output_path = PROJECT_DIR / 'data' / 'games_engineered.csv'
games.to_csv(output_path, index=False)
print('File tersimpan:', output_path)
```
